```python
df = pd.read_csv('/content/drive/MyDrive/ML Practice Heart Disease Logistic Regression/heart failure prediction.csv')
df.head()
df.info()
```

```python
#To check/find missing values from dataset
df.isnull().sum()
```

```python
# The 'sklearn.preprocessing'for check any 'object'type data.
categorical_cols = df.select_dtypes(include='object').columns

for col in categorical_cols:
    print(f"Unique values for '{col}':")
    display(df[col].unique())
    print("\n")
```

```python
# One-Hot Encoding: This process converts categorical text data into a numerical format
# (0s and 1s) that machine learning models can understand. 'drop_first=True'
# helps prevent multicollinearity by dropping one dummy variable per category.

# Identify all columns that still have 'object' (string) data type
columns_to_encode = df.select_dtypes(include='object').columns.tolist()

if columns_to_encode:
    print(f"One-hot encoding the following columns: {columns_to_encode}")
    df = pd.get_dummies(df, columns=columns_to_encode, drop_first=True)
else:
    print("All categorical columns have already been one-hot encoded.")
```

```python
df.columns
```

```python
df.isnull().sum()
```

```python
# The 'df' DataFrame should already be prepared with all categorical columns one-hot encoded
# by the previous data preprocessing steps.

x = df.drop('HeartDisease', axis=1)
y = df['HeartDisease']

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

model = RandomForestClassifier(random_state=42)

model.fit(x_train, y_train)

y_pred = model.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix

# Calculate evaluation metrics
accuracy = accuracy_score(y_test, y_pred)#Accuracy batati hai model ne total predictions mein se kitni predictions sahi ki.Agar Training Accuracy bohot high ho aur Testing Accuracy bohot low ho, to high Training Accuracy koi achievement nahi hai.
precision = precision_score(y_test, y_pred)#"Jin logon ko model ne Disease bola, unmein se kitne waqai Disease wale the"
recall = recall_score(y_test, y_pred)#Asal Disease walon mein se kitnon ko machine ne pakra?
f1 = f1_score(y_test, y_pred)#F1-Score = Precision aur Recall ka balanced report card.
conf_matrix = confusion_matrix(y_test, y_pred)

print(f"Accuracy: {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall: {recall:.4f}")
print(f"F1-score: {f1:.4f}")
print("\nConfusion Matrix:")
display(pd.DataFrame(conf_matrix, index=['Actual 0', 'Actual 1'], columns=['Predicted 0', 'Predicted 1']))
```
